### Що з переліку є AI?

## Завдання 2. Правила для підозрілого оголошення OLX

Склав чотири правила. Оголошення потребує перевірки, якщо виконується хоча б одне:

1. Ціна нижча за половину орієнтовної ринкової ціни.
2. Продавець вимагає повну передоплату до отримання товару.
3. Акаунту менше семи днів і продавець наполягає перейти у сторонній месенджер.
4. Продавець просить повідомити код підтвердження банківської операції з SMS.

Сигнал цих правил не є доказом шахрайства. Для перевірки використовую чотири
вигадані оголошення з наперед заданою позначкою «шахрайство / ні».

```python
import pandas as pd

pd.set_option("display.max_colwidth", None)

def check_ad(ad):
    if ad["ціна"] < ad["ринкова_ціна"] * 0.5:
        return True
    if ad["повна_передоплата"]:
        return True
    if ad["вік_акаунта"] < 7 and ad["сторонній_месенджер"]:
        return True
    if ad["просить_код_sms"]:
        return True
    return False

ads = [
    {"назва": "Звичайне оголошення", "ціна": 9000, "ринкова_ціна": 10000,
     "повна_передоплата": False, "вік_акаунта": 2, "сторонній_месенджер": False,
     "просить_код_sms": False, "шахрайство": False},
    {"назва": "Підозріло низька ціна", "ціна": 3000, "ринкова_ціна": 10000,
     "повна_передоплата": True, "вік_акаунта": 2, "сторонній_месенджер": True,
     "просить_код_sms": False, "шахрайство": True},
    {"назва": "Шахрайство без ознак правил", "ціна": 9500, "ринкова_ціна": 10000,
     "повна_передоплата": False, "вік_акаунта": 600, "сторонній_месенджер": False,
     "просить_код_sms": False, "шахрайство": True},
    {"назва": "Чесний розпродаж", "ціна": 4500, "ринкова_ціна": 10000,
     "повна_передоплата": False, "вік_акаунта": 500, "сторонній_месенджер": False,
     "просить_код_sms": False, "шахрайство": False},
]

results = []
for ad in ads:
    signal = check_ad(ad)
    results.append({
        "Приклад": ad["назва"],
        "Справді шахрайство": ad["шахрайство"],
        "Сигнал правил": signal,
        "Правильний висновок": signal == ad["шахрайство"],
    })

pd.DataFrame(results)
```

### Що показала перевірка

- У другому прикладі правила виявили підозріле оголошення.
- Третій приклад обійшов усі правила: ціна звичайна, акаунт давній, небезпечних
  запитів в оголошенні немає. Обман може з'ясуватися пізніше, наприклад після
  отримання товару, який не відповідає опису.
- Четвертий приклад чесний, але знижка спричинила хибне спрацювання.
- Чотирьох правил недостатньо. Навіть довший список не охопить усі ситуації.
  Через місяць способи обману можуть змінитися, а самі правила не навчаться нічому новому.

**Які дані потрібні для ML:** історія оголошень, ціни та категорії товарів,
текст і фотографії, вік акаунта, попередні оголошення продавця, скарги та результати
перевірки модераторів. Потрібна правильна цільова позначка: підтверджене шахрайство
або чесне оголошення. Сама скарга ще не доводить шахрайство.

Якщо перевіряємо оголошення під час публікації, майбутні скарги й рішення модератора
можна використати як джерело позначок для старих прикладів, але не як вхідні ознаки
для нового оголошення. На момент прогнозу цих даних ще немає.

**Висновок:** прості правила корисні для первинної перевірки. ML шукає складніші
закономірності на позначених прикладах, але також може помилятися і потребує перевірки.

### Бліц: правила чи ML?

| Задача | Вибір | Пояснення |
| --- | --- | --- |
| Перевести гривні в долари за відомим курсом НБУ | Правила | Достатньо точної формули. |
| Розпізнати рукописний індекс | ML | Почерк має багато варіантів. |
| Спрогнозувати попит на обігрівачі | ML | Потрібні закономірності продажів, сезону та погоди. |
| Визначити складання іспиту за порогом 60 балів | Правила | Достатньо порівняти бал із порогом. |
| Визначити позитивний чи негативний відгук | ML | Потрібно врахувати зміст, сленг і контекст. |

## Завдання 4. Розбір датасету смартфонів Rozetka

У завданні наведено лише назви 12 стовпців, без реальних рядків товарів.
Нижче класифікую саме ці ознаки, не підставляючи вигадані дані Rozetka.

### Тип кожної ознаки

```python
features = pd.DataFrame([
    ["бренд", "Категоріальна номінальна", "Назви без природного порядку."],
    ["модель", "Категоріальна номінальна", "Назва / ідентифікатор моделі, не числова величина."],
    ["ціна_грн", "Числова неперервна", "Грошову величину умовно розглядаємо як неперервну."],
    ["пам’ять_гб", "Числова дискретна", "Задані окремі значення: 64, 128, 256, 512."],
    ["діагональ", "Числова неперервна", "Вимірюється, наприклад 6.1 або 6.7 дюйма."],
    ["рік_виходу", "Числова дискретна", "Рік задано цілим числом."],
    ["колір", "Категоріальна номінальна", "Кольори не мають природного порядку."],
    ["підтримка_5G", "Бінарна", "Лише так або ні."],
    ["рейтинг_відгуків", "Категоріальна порядкова", "Оцінки 1–5 впорядковані; рівність інтервалів не гарантована."],
    ["кількість_продажів_за_місяць", "Числова дискретна", "Рахуємо продані одиниці цілим числом."],
    ["текст_першого_відгуку", "Текстова", "Довільний текст покупця."],
    ["чи_був_повернений_покупцем", "Бінарна", "Лише так або ні."],
], columns=["Ознака", "Тип", "Пояснення"])

features
```

### Одна задача регресії

**Питання:** скільки одиниць нового смартфона продадуть протягом першого місяця?

- **Цільова ознака `y`:** `кількість_продажів_за_місяць`.
- **Вхідні ознаки `X`:** бренд, модель, стартова ціна, пам'ять, діагональ, рік виходу,
  колір і підтримка 5G — лише значення, відомі до старту продажів.
- **Кому корисно:** магазину для планування закупівель і запасів.
- **Чому регресія:** прогнозуємо кількість, а не назву класу. Ціль може бути цілим числом.

Навчальні приклади мають містити продажі за однаковий період — перший місяць,
а не суміш першого місяця й довільного місяця після виходу.

### Одна задача класифікації

**Питання:** чи поверне покупець придбаний смартфон?

- **Цільова ознака `y`:** `чи_був_повернений_покупцем` — так / ні.
- **Вхідні ознаки `X`:** відомі перед купівлею характеристики та ціна смартфона.
- **Кому корисно:** магазину для перевірки опису товару й планування роботи з поверненнями.
- **Чому класифікація:** вибираємо один із двох класів.

Для такого прогнозу один рядок повинен відповідати одній покупці. Якщо вихідна
таблиця описує моделі смартфонів, а позначка означає лише «хоч раз був повернений»,
вона не дає відповіді про повернення конкретним покупцем. Для цього потрібна
історія окремих покупок і їхніх повернень. На рівні моделей можна класифікувати
лише наявність хоча б одного повернення за однаковий заданий період.

### Витік даних: що не можна брати до старту продажів

Для прогнозу продажів нового смартфона ще немає:

1. `рейтинг_відгуків` — з'являється після відгуків покупців.
2. `текст_першого_відгуку` — з'являється після першого відгуку.
3. `чи_був_повернений_покупцем` — результат відомий після покупки та можливого повернення.

Використання цих майбутніх значень завищить якість перевірки. На новому смартфоні
модель не отримає таких даних. Стовпець продажів також не входить до `X`, бо він є
самою відповіддю `y`. Ціна повинна бути стартовою, а не ціною після майбутніх знижок.

```python
target = "кількість_продажів_за_місяць"
unavailable = [
    "рейтинг_відгуків",
    "текст_першого_відгуку",
    "чи_був_повернений_покупцем",
]

input_features = []
for feature in features["Ознака"]:
    if feature != target and feature not in unavailable:
        input_features.append(feature)

pd.DataFrame({"Ознаки X, відомі до старту продажів": input_features})
```
